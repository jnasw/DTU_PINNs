```python
from src.dataset.create_dataset_functions import ODE_modelling

# Create model instance
ode_model = ODE_modelling(config)

# Load dataset
dataset = ode_model.load_dataset("dataset_v1.pkl")  # Change name if needed

print(f"Loaded {len(dataset)} trajectories")
```

```python
import plotly.graph_objs as go
import plotly.io as pio

# Load single trajectory
i = 0
trajectory = dataset[i]
t = trajectory[0]  # Time vector
states = trajectory[1:]

# State variable labels
labels = ['θ', 'ω', 'E_d′', 'E_q′', 'R_F', 'V_r', 'E_fd', 'P_sv', 'P_m']

# Create interactive figure
fig = go.Figure()

for idx, state in enumerate(states):
    fig.add_trace(go.Scatter(
        x=t,
        y=state,
        mode='lines',
        name=labels[idx]
    ))

fig.update_layout(
    title=f"Simulated Trajectory {i} — Synchronous Machine (SM_AVR_GOV)",
    xaxis_title="Time [s]",
    yaxis_title="State Values",
    hovermode='x unified',
    template='plotly_white',
    legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1)
)

fig.show()
```
